# 🎙️ OmniVoice TTS Server on Google Colab (GPU)

Notebook này biến Google Colab thành **GPU TTS Server**, cho phép máy Mac của bạn gửi file markdown lên để sinh audio với tốc độ siêu nhanh (nhờ GPU T4/L4/A100) và tự động tải file thành phẩm về máy Mac.

### Hướng dẫn sử dụng:
1. Chọn **Runtime** $\rightarrow$ **Change runtime type** $\rightarrow$ Chọn **T4 GPU** (hoặc A100 nếu có Pro).
2. Chạy lần lượt các ô (cells) bên dưới.
3. Khi ô cuối cùng chạy, copy đường link `https://xxxx.trycloudflare.com` hiển thị trên màn hình.
4. Trên Terminal máy Mac, lưu server URL bằng lệnh:
   ```bash
   ./ov-remote --set-server https://xxxx.trycloudflare.com
   ```
5. Bắt đầu sinh audio từ máy Mac:
   ```bash
   ./ov-remote input.md
   ```
